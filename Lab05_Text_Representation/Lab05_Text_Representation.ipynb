# 1- TF-IDF

<img src="https://drive.google.com/uc?export=view&id=1saSt0mMgOQ2ybbTms1lu_ruhUcBWkKzL" width="500">

TF-IDF stands for term frequency-inverse document frequency. It is a measure that discounts common words. Used in the fields of information retrieval (IR) and machine learning, that can quantify the importance of words in a document amongst a collection of documents (also known as a corpus).

### Components of TF-IDF

1. **TF (Term Frequency):**  
   Measures how frequently a term appears in a document.


$$ \text{TF}(t, d) = \frac{\text{Number of times term } t \text{ appears in document } d}{\text{Total number of terms in document } d} $$


2. **IDF (Inverse Document Frequency):**  
   Measures how important a term is across all documents. Words that appear in many documents get lower scores.

$$    \text{IDF}(t) = \log \left(\frac{\text{Number of all documents N}}{\text{Number of documents containing the term } t}\right) $$

# Cosine similarity

In NLP, Cosine similarity is a metric used to measure how similar the documents are.


$$ \text{cosine similarity} = \frac{A \cdot B}{\|A\| \times \|B\|} $$

Where:  
$ A \cdot B $  = dot product of vectors A and B  
$ \|A\| $ = magnitude (length) of vector A  
$ \|B\| $ =  magnitude (length) of vector B

#### Intuition

- If vectors point in the **same direction**, cosine similarity = **1** (maximum similarity).
- If vectors are **orthogonal (90° apart)**, cosine similarity = **0** (no similarity).

Word2Vec utilizes two architectures :

### Task 1: Cosine Similarity
Use the Cosine Similarity method to determine how similar the following sentences are.

'This is the first document.',  
'This document is the second document.',  
'And this is the third one.',  
'Is this the first document?'  


````python
documents = [
    'This is the first document.',
    'This document is the second document.',
    'And this is the third one.',
    'Is this the first document?',
]

vectorizer = TfidfVectorizer()
doc_vectors = vectorizer.fit_transform(documents)

similarity = cosine_similarity(doc_vectors)
pd.DataFrame(similarity, index=documents, columns=documents).round(3)
````

````python
# Rank the sentence pairs from most to least similar
pairs = [(documents[i], documents[j], similarity[i, j])
         for i in range(len(documents)) for j in range(i + 1, len(documents))]
for a, b, s in sorted(pairs, key=lambda p: -p[2]):
    print(f"{s:.3f}  {a!r:42} <-> {b!r}")
````

**Observation:** *"This is the first document."* and *"Is this the first document?"* have a cosine similarity of **1.0** –
after lowercasing and removing punctuation they contain exactly the same words (TF-IDF ignores word order).
The third sentence (*"And this is the third one."*) shares only common words such as *this, is, the*, so it is the least similar to the others.

### Task 2: TF-IDF
Use tf-idf method on the sentences below to determine the important words.

'data science is one of the most important fields of science',  
'this is one of the best data science courses',  
'data scientists analyze data'  


````python
sentences = [
    'data science is one of the most important fields of science',
    'this is one of the best data science courses',
    'data scientists analyze data',
]

tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(sentences)

df_tfidf = pd.DataFrame(tfidf_matrix.toarray(), columns=tfidf_vectorizer.get_feature_names_out(),
                        index=[f"sentence {i+1}" for i in range(len(sentences))])
df_tfidf.round(3)
````

````python
# The most important (highest TF-IDF) words in each sentence
for name, row in df_tfidf.iterrows():
    top = row.sort_values(ascending=False).head(3)
    print(name, "->", ", ".join(f"{w} ({s:.3f})" for w, s in top.items()))

# idf: words that appear in every sentence (like 'data') get the lowest idf
idf = pd.Series(tfidf_vectorizer.idf_, index=tfidf_vectorizer.get_feature_names_out()).sort_values()
print("\nIDF values:\n", idf.round(3).to_string())
````

**Observation:** `data` appears in all three sentences, so it has the lowest IDF (1.0). Words that appear in only one sentence
(*best, courses, this, fields, important, most, analyze, scientists*) have the highest IDF (1.693).
- Sentence 1 → **science** and **of** (each used twice) and **fields / important / most**.
- Sentence 2 → **best, courses, this** – the words unique to this sentence.
- Sentence 3 → **data** (it appears twice in this short sentence, so its TF is high even though its IDF is low), **analyze, scientists**.

So TF-IDF rewards words that are frequent in a document but rare across documents. Note that stop words such as *of* can still score high;
removing them (`TfidfVectorizer(stop_words="english")`) would leave only content words.

````python
import re
import gensim

simpsons = pd.read_csv("dataset/simpsons_script_lines.csv", usecols=["spoken_words"],
                       on_bad_lines="skip", engine="python")
simpsons = simpsons.dropna(subset=["spoken_words"])
print("Lines with spoken words:", len(simpsons))
simpsons.head()
````

````python
def clean_text(text):
    text = text.lower()
    text = re.sub(r"[0-9]", '', text)
    text = re.sub(r"[)(,”“.’$-]", '', text)
    return text

simpsons["clean"] = simpsons["spoken_words"].apply(clean_text)

# Tokenize every line (each line = one "sentence" for Word2Vec)
tokens = [word_tokenize(line) for line in simpsons["clean"]]
print(tokens[:3])
````

````python
Skip_gram_model = gensim.models.Word2Vec(tokens, min_count = 1, vector_size = 100, window = 5, sg = 1, seed = 42)
print("Vocabulary size:", len(Skip_gram_model.wv))
````

### Task 4

Use: wv.most_similar() method to :

1.	Find the words similar to “homer”.

2.	Find the words similar to “marge”.

3. Find the words similar to “bart”



````python
for name in ["homer", "marge", "bart"]:
    print(f"Words most similar to '{name}':")
    for word, score in Skip_gram_model.wv.most_similar(name, topn=10):
        print(f"   {word:15} {score:.3f}")
    print()
````
