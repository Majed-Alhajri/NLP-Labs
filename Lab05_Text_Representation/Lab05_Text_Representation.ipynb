# 1- TF-IDF

<img src="https://drive.google.com/uc?export=view&id=1saSt0mMgOQ2ybbTms1lu_ruhUcBWkKzL" width="500">

TF-IDF stands for term frequency-inverse document frequency. It is a measure that discounts common words. Used in the fields of information retrieval (IR) and machine learning, that can quantify the importance of words in a document amongst a collection of documents (also known as a corpus).

### Components of TF-IDF

1. **TF (Term Frequency):**  
   Measures how frequently a term appears in a document.


$$ \text{TF}(t, d) = \frac{\text{Number of times term } t \text{ appears in document } d}{\text{Total number of terms in document } d} $$


2. **IDF (Inverse Document Frequency):**  
   Measures how important a term is across all documents. Words that appear in many documents get lower scores.

$$    \text{IDF}(t) = \log \left(\frac{\text{Number of all documents N}}{\text{Number of documents containing the term } t}\right) $$

# Cosine similarity

In NLP, Cosine similarity is a metric used to measure how similar the documents are.


$$ \text{cosine similarity} = \frac{A \cdot B}{\|A\| \times \|B\|} $$

Where:  
$ A \cdot B $  = dot product of vectors A and B  
$ \|A\| $ = magnitude (length) of vector A  
$ \|B\| $ =  magnitude (length) of vector B

#### Intuition

- If vectors point in the **same direction**, cosine similarity = **1** (maximum similarity).
- If vectors are **orthogonal (90° apart)**, cosine similarity = **0** (no similarity).

### Task 1: Cosine Similarity
Use the Cosine Similarity method to determine how similar the following sentences are.

'This is the first document.',  
'This document is the second document.',  
'And this is the third one.',  
'Is this the first document?'  


````python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = ['This is the first document.',
             'This document is the second document.',
             'And this is the third one.',
             'Is this the first document?']

tfidf_vectorizer = TfidfVectorizer()
vector_matrix = tfidf_vectorizer.fit_transform(documents)

similarity = cosine_similarity(vector_matrix)
df_similarity = pd.DataFrame(similarity, index=documents, columns=documents).round(3)
df_similarity
````

**Result:**
- *"This is the first document."* and *"Is this the first document?"* have a cosine similarity of **1.0**. After tokenization and lower-casing they contain exactly the same words (`this, is, the, first, document`). TF-IDF ignores word order and punctuation, so their vectors are identical.
- *"This document is the second document."* is quite similar to them (**0.647**): it shares *this, is, the, document*, and *document* appears twice.
- *"And this is the third one."* is the least similar to the others (**0.225 – 0.308**): it only shares the common words *this, is, the*.

### Task 2: TF-IDF
Use tf-idf method on the sentences below to determine the important words.

'data science is one of the most important fields of science',  
'this is one of the best data science courses',  
'data scientists analyze data'  


````python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer

sentences = ['data science is one of the most important fields of science',
             'this is one of the best data science courses',
             'data scientists analyze data']

tfidf = TfidfVectorizer()
result = tfidf.fit_transform(sentences)

df_tfidf = pd.DataFrame(result.toarray(), columns=tfidf.get_feature_names_out(),
                        index=['Sentence 1', 'Sentence 2', 'Sentence 3']).round(3)
df_tfidf
````

````python
# Most important (highest TF-IDF) words in each sentence
for i, row in df_tfidf.iterrows():
    top = row.sort_values(ascending=False).head(3)
    print(f"{i}: " + ", ".join(f"{w} ({s})" for w, s in top.items()))
````

**Important words** (highest TF-IDF in each sentence):
- **Sentence 1:** *science* (0.488, appears twice) and *fields*, *important*, *most* (0.321, unique to this sentence). *of* also scores high only because it appears twice. A stop-word list (`TfidfVectorizer(stop_words='english')`) would remove it.
- **Sentence 2:** *best*, *courses*, *this* (0.4) — words that occur only in this sentence.
- **Sentence 3:** *data* (0.641, appears twice), *analyze* and *scientists* (0.543).

Words that appear in **every** sentence (e.g. *data*) get a lower IDF. They become important only where they are repeated many times (*data* in sentence 3). Words unique to one sentence get the highest weights.

````python
import re
import gensim
import pandas as pd

simpsons = pd.read_csv('dataset/simpsons_script_lines.csv', engine='python', on_bad_lines='skip')
simpsons = simpsons.dropna(subset=['spoken_words'])
print("Number of lines:", len(simpsons))

def clean_text(text):
    text = text.lower()
    text = re.sub(r"[0-9]", '', text)
    text = re.sub(r"[)(,”“.’$-]", '', text)
    return text

simpsons['clean_words'] = simpsons['spoken_words'].astype(str).apply(clean_text)

# tokenize each line into a list of words
tokens = [line.split() for line in simpsons['clean_words'] if line.strip()]
print("Example tokens:", tokens[:3])

Skip_gram_model = gensim.models.Word2Vec(tokens, min_count=1, vector_size=100, window=5, sg=1, seed=42)
print("Vocabulary size:", len(Skip_gram_model.wv))
````

### Task 4

Use: wv.most_similar() method to :

1.	Find the words similar to “homer”.

2.	Find the words similar to “marge”.

3. Find the words similar to “bart”



````python
for name in ['homer', 'marge', 'bart']:
    print(f"Words similar to '{name}':")
    for word, score in Skip_gram_model.wv.most_similar(name, topn=10):
        print(f"   {word:15} {score:.3f}")
    print()
````
